```python
#[CELL 1] STUDENT CONFIGURATION
LAST_NAME = "Obiedo"
STUDENT_ID = "TUPM-26-4134"

SEED_DIGIT = int(STUDENT_ID[-1])  
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)

```

```python
# TASK 6 
#Step 1: Create a dataset
data = [SEED_DIGIT, ID_SUM, NAME_LENGTH] # Make a Dataset (Input)

#Step 2: Find the code to find the largest value in the dataset and use it for processing.
largest_value = max(data) # Find the largest value in the dataset (Processing)

#Step 3: Print the result of processing. 
print("Data:", data) # Output: List of Data in the Dataset
print("Largest Value:", largest_value) #Output: Largest Value in the Dataset (Output)
```

```python
# [CELL 2] LARGEST VALUE ALGORITHM
data = [ 
    10 + SEED_DIGIT,
    5 + NAME_LENGTH,
    20 + SEED_DIGIT,
    8 + NAME_LENGTH,
    15 + ID_SUM
]

largest = data[0] 
comparisons = 0

for value in data[1:]:
    comparisons += 1
    if value > largest:
        largest = value

print("Data:", data)
print("Largest:", largest)
print("Comparisons:", comparisons)
```

```python
# [CELL 3] LINEAR GROWTH
def linear_operations(n):
    count = 0 
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
    
```
