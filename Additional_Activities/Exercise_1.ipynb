```python
#[CELL 1] Student information  
LAST_NAME = "OBIEDO"
STUDENT_ID = "TUPM-26-4134"
FAVORITE_ARTIST = "CUP OF JOE"
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(FAVORITE_ARTIST.lower()) - 96 for FAVORITE_ARTIST in FAVORITE_ARTIST if FAVORITE_ARTIST.isalpha())

print ("Student ID:", STUDENT_ID)
print ("Seed DIGIT:", SEED_DIGIT)
print ("Favorite Artist:", FAVORITE_ARTIST)
print ("ID sum:", ID_SUM)
print ("Name Length:", NAME_LENGTH)
print ("Artist Length:", ARTIST_LENGTH)
print ("Artist Sum:", ARTIST_SUM)
```

```python
#[CELL 2] data sets
records = [ID_SUM + i * (NAME_LENGTH + 2) for i in range(12)]
rotation = SEED_DIGIT % len(records)
rotated = records[rotation:] + records[:rotation]

target = records[(ID_SUM + ARTIST_SUM) % len(records)]   

print("Original Records:", records)
print("Rotation Point:", rotation)
print("Rotated Records:", rotated)
print("Target Incident:", target)
```

```python
#[CELL 3] Searches (both algorithms are from the manual)
def linear_search(data, target):          # Task 12
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

def binary_search(data, target):          # Task 13, unchanged
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

# Why plain Binary Search alone fails on rotated data
print("Plain Binary on rotated list:", binary_search(rotated, target))

# Binary Search on the correct sorted piece
boundary = len(rotated) - rotation        # where the moved records begin
if target >= rotated[0]:
    idx_r, comp_r = binary_search(rotated[:boundary], target)
else:
    idx_r, comp_r = binary_search(rotated[boundary:], target)
    if idx_r != -1:
        idx_r += boundary                 # convert back to rotated-list position

idx_l, comp_l = linear_search(rotated, target)

print(f"\nBinary Search (sorted piece) -> index={idx_r}, comparisons={comp_r}")
print(f"Linear Search               -> index={idx_l}, comparisons={comp_l}")
print(f"Comparison Reduction: {comp_l - comp_r}")
print("Time Complexity: O(log n) | Space Complexity: O(n)")
```

```python
#[CELL 4] Final Output
if idx_r != -1:
    print(f"Target Incident {target} found at index {idx_r} of the rotated records.")
else:
    print(f"Target Incident {target} not found in the rotated records.")
print(f"Binary Search: {comp_r} comparisons | Linear Search: {comp_l} comparisons | Reduction: {comp_l - comp_r}")
print("Time Complexity: O(log n) | Space Complexity: O(n)")
```
