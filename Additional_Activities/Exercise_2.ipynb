```python
#[CELL 1] Student information  
LAST_NAME = "OBIEDO"
STUDENT_ID = "TUPM-26-4134"
FAVORITE_ARTIST = "CUP OF JOE"
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(FAVORITE_ARTIST.lower()) - 96 for FAVORITE_ARTIST in FAVORITE_ARTIST if FAVORITE_ARTIST.isalpha())

print ("Student ID:", STUDENT_ID)
print ("Seed DIGIT:", SEED_DIGIT)
print ("Favorite Artist:", FAVORITE_ARTIST)
print ("ID sum:", ID_SUM)
print ("Name Length:", NAME_LENGTH)
print ("Artist Length:", ARTIST_LENGTH)
print ("Artist Sum:", ARTIST_SUM)
```

```python
#[CELL 2] Data Sets 
#DATA SET A
dataset_a = [(i + 1) * SEED_DIGIT + ID_SUM for i in range(8)]
 
#DATA SET B
dataset_b = [(i + 1) * ARTIST_LENGTH + NAME_LENGTH for i in range(8)]

#Output
print("Department A Dataset:", dataset_a)
print("Department B Dataset:", dataset_b)
```

```python
#[CELL 3] Insertion Sort Consolidation
def insertion_sort(data):                 # Task 17
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

combined = dataset_a + dataset_b
consolidated, comparisons, shifts = insertion_sort(combined)

print("Department A Dataset:", dataset_a)
print("Department B Dataset:", dataset_b)
print("Consolidated Dataset:", consolidated)
print(f"Records Before: {len(dataset_a)} + {len(dataset_b)} = {len(combined)} | After: {len(consolidated)}")
print("Comparison Count:", comparisons)
print("Shifts:", shifts)
print("Ordering Verified:", consolidated == sorted(consolidated))
print("Time Complexity: O(n²) worst case, n = total records | Space Complexity: O(n)")
```
