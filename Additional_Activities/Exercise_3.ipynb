```python
#[CELL 1] Student information  
LAST_NAME = "OBIEDO"
STUDENT_ID = "TUPM-26-4134"
FAVORITE_ARTIST = "CUP OF JOE"
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(FAVORITE_ARTIST.lower()) - 96 for FAVORITE_ARTIST in FAVORITE_ARTIST if FAVORITE_ARTIST.isalpha())

print ("Student ID:", STUDENT_ID)
print ("Seed DIGIT:", SEED_DIGIT)
print ("Favorite Artist:", FAVORITE_ARTIST)
print ("ID sum:", ID_SUM)
print ("Name Length:", NAME_LENGTH)
print ("Artist Length:", ARTIST_LENGTH)
print ("Artist Sum:", ARTIST_SUM)
```

```python
#[CELL 2] Bubble sort
import time

def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps
```

```python
#[CELL 3] Selection Sort
def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps
```

```python
#[CELL 4] Insertion Sort
def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts
```

```python
#[CELL 5] Sorting Comparison
order_count = NAME_LENGTH + ARTIST_LENGTH
orders = [(ID_SUM * 3 + i * ARTIST_SUM) % 500 for i in range(order_count)]

conditions = {
    "Sorted": sorted(orders),
    "Reverse": sorted(orders, reverse=True),
    "Random": orders,
}
algorithms = [("Bubble", bubble_sort), ("Selection", selection_sort), ("Insertion", insertion_sort)]

print("Generated Orders:", orders)
for cond_name, dataset in conditions.items():
    print(f"{cond_name} Input:", dataset)

results = {}
for cond_name, dataset in conditions.items():
    print(f"\n--- Condition: {cond_name} ---")
    for algo_name, algo in algorithms:
        start = time.perf_counter()
        result, comparisons, moves = algo(dataset)
        elapsed = time.perf_counter() - start
        verified = result == sorted(dataset)
        results[(cond_name, algo_name)] = (comparisons, moves, elapsed, verified)
        print(f"{algo_name:<10} | comparisons={comparisons:4d} | movements={moves:4d} "
              f"| time={elapsed:.8f}s | sorted correctly={verified}")
```

```python
#[CELL 6] Analysis and Final Output
print("--- Big O Analysis ---")
big_o = {
    "Bubble": ("O(n)", "O(n²)", "O(n²)"),
    "Selection": ("O(n²)", "O(n²)", "O(n²)"),
    "Insertion": ("O(n)", "O(n²)", "O(n²)"),
}
print(f"{'Algorithm':<10} | {'Best':<6} | {'Average':<7} | {'Worst':<6} | Space")
for algo_name, (best, average, worst) in big_o.items():
    print(f"{algo_name:<10} | {best:<6} | {average:<7} | {worst:<6} | O(n)")

print("\n--- Best-, Average-, and Worst-Case Observations ---")
case_map = [("Best (Sorted)", "Sorted"), ("Average (Random)", "Random"), ("Worst (Reverse)", "Reverse")]
for label, cond_name in case_map:
    print(label)
    for algo_name, _ in algorithms:
        comparisons, moves, elapsed, verified = results[(cond_name, algo_name)]
        print(f"  {algo_name:<10} comparisons={comparisons:4d} | movements={moves:4d}")

print("\n--- Final Output ---")
print("Orders Processed:", order_count)
print("All Results Sorted Correctly:", all(r[3] for r in results.values()))
for cond_name in conditions:
    def leaders(index):
        lowest = min(results[(cond_name, name)][index] for name, _ in algorithms)
        return " / ".join(name for name, _ in algorithms if results[(cond_name, name)][index] == lowest)
    print(f"{cond_name:<8} | Fewest Comparisons: {leaders(0)} | Fewest Movements: {leaders(1)} | Fastest: {leaders(2)}")
```
